# Run BentoML Service Notebook
This notebook explains how to start the BentoML service and verify the API endpoints.


## 1. Start the service
Run the BentoML service in a separate terminal, because the server process stays active.


```python
print("Run this command in another terminal:")
print("bentoml serve seattle_energy.service:EnergyService --port 3000")
```

## 2. Health check
Once the service is running, use `/ping` to confirm the API is ready.


```python
import requests
base_url = "http://localhost:3000"
response = requests.get(f"{base_url}/ping")
print("Status:", response.status_code)
print(response.json())
```

## 3. Prediction request
Use the following payload template to call `/predict`. Each `Use_*` field is a binary flag (0 or 1).


```python
payload = {
    "log_surface": 8.5,
    "percent_electricity": 70.0,
    "has_parking": 1,
    "BuildingAge": 40,
    "surface_per_floor": 5000.0,
    "Use_Hotel": 0,
    "Use_Office": 1,
    "Use_Retail_Store": 0,
    "Use_Other": 0,
    "Use_Non_Refrigerated_Warehouse": 0,
    "Use_K12_School": 0,
    "Use_Medical_Office": 0,
    "Use_Worship_Facility": 0,
    "Use_Unknown": 0
}
response = requests.post(f"{base_url}/predict", json=payload)
print("Status:", response.status_code)
print(response.json())
```
