# Run BentoML Service Notebook
This notebook replaces `scripts/run_service.py`. It explains how to start the BentoML service and how to send requests once the API is running.

## Start the service
Run the BentoML service in a separate terminal because the server command keeps running.

```bash
bentoml serve seattle_energy.service:EnergyService --port 3000
```

## Example request from Python
Use the next cell only after the service is running on port 3000.

````python
import requests

payload = {
    "log_surface": 8.5,
    "percent_electricity": 70.0,
    "has_parking": 1,
    "BuildingAge": 40,
    "surface_per_floor": 5000.0,
    "Use_Office": 1,
    "Use_Hotel": 0,
    "Use_Retail_Store": 0,
    "Use_Other": 0,
    "Use_Non_Refrigerated_Warehouse": 0,
    "Use_K12_School": 0,
    "Use_Medical_Office": 0,
    "Use_Worship_Facility": 0,
    "Use_Unknown": 0
}
response = requests.post("http://localhost:3000/predict", json=payload)
print(response.status_code)
print(response.json())
````
