# Prepare Data Notebook
This notebook combines data loading, exploratory analysis, and the feature engineering pipeline. It is designed to illustrate the process in a pedagogical way, while keeping the structure simple and reusable.


```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
from seattle_energy.data_processing import load_raw_data, build_feature_dataframe, save_processed_data, RAW_DATA_PATH

print(f"Raw data path: {RAW_DATA_PATH}")
raw = load_raw_data()
print(f"Loaded raw dataset with {len(raw)} rows and {len(raw.columns)} columns")
```

```python
raw.head()
```

## First exploration
Check column names, null counts, and object-type categories before applying the main pipeline.


```python
print("Columns:
", raw.columns.tolist())
print("
Missing values by column:
")
missing = raw.isnull().sum().sort_values(ascending=False)
print(missing[missing > 0])
```

```python
cat_cols = raw.select_dtypes(include=["object"]).columns.tolist()
print("Object columns (categorical/text):", cat_cols)
for col in cat_cols[:6]:
    print(f"
{col}: {raw[col].nunique()} unique values")
```

## Target and feature view
Inspect key numeric columns used for energy prediction, and verify that values are valid for modeling.


```python
numeric = raw.select_dtypes(include=["number"]).columns.tolist()
print("Numeric columns used for physics and energy metrics:
", numeric)
raw[["SiteEnergyUse(kBtu)", "Electricity(kBtu)", "PropertyGFABuilding(s)", "YearBuilt", "NumberofFloors"]].describe().T
```

## Filter and build features
Use the pipeline function from `seattle_energy/data_processing.py` to create a clean dataset for modeling.


```python
processed = build_feature_dataframe(raw)
print(f"Processed dataset has {len(processed)} rows and {len(processed.columns)} columns")
processed.head()
```

```python
save_processed_data(processed)
print("Saved processed dataset to data/processed/feature_engineered_cleaned_for_bento.csv")
```
